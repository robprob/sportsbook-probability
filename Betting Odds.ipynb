```python
class American:
    def __init__(self, A):
        if A == 0:
            raise Exception('American odds cannot be zero.')
        self.odds = float(A)

    def prob(self):
        """
        Calculate underlying implied probability.
        """
        A = self.odds
        if A < 0:
            return -A / (-A + 100.0)
        elif A > 0:
            return 100 / (A + 100.0)

    def payout(self, bet):
        """
        Calculate total winning payout scaled to initial bet.
        """
        A = self.odds
        if A < 0:
            # Scale number of wins to bet amount
            num_wins = bet / abs(A)
            # Wins are scaled around $100
            payout = num_wins * 100
            # Add initial bet amount
            payout += bet
        elif A > 0:
            # Scale number of wins to bet amount
            num_wins = bet / 100.0
            # Wins are scaled around odds
            payout = num_wins * A
            # Add initial bet amount
            payout += bet

        return payout

    def profit(self, bet):
        """
        Calculate winning net profit given initial bet.
        """
        return self.payout(bet) - bet
```

```python
class Decimal:
    def __init__(self, D):
        if D <= 1:
            raise Exception('Decimal odds must be greater than 1.')
        self.odds = float(D)

    def prob(self):
        """
        Calculate underlying implied probability.
        """
        return 1 / self.odds

    def payout(self, bet):
        """
        Calculate total winning payout scaled to initial bet.
        """
        return self.odds * bet

    def profit(self, bet):
        """
        Calculate winning net profit given initial bet.
        """
        return (self.odds * bet) - bet
```

```python
class Fractional:
    def __init__(self, F):
        if F <= 0:
            raise Exception('Fractional odds cannot be zero or negative.')
        self.odds = float(F)

    def prob(self):
        """
        Calculate underlying implied probability.
        """
        return 1 / (self.odds + 1)

    def payout(self, bet):
        """
        Calculate total winning payout scaled to initial bet.
        """
        return (self.odds * bet) + bet

    def profit(self, bet):
        """
        Calculate winning net profit given initial bet.
        """
        return self.odds * bet
```
